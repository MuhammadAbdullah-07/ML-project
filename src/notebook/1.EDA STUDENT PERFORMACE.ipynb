### SUM OF COLUMNS AND AVG OF COLUMNS

```python
df['Total_Score'] = df['math score'] + df['reading score'] + df['writing score']
df['Avg_Score'] = df['Total_Score'] / 3
```

```python
df

```

```python
reading_full=df[df['reading score']==100]['Avg_Score'].count()
writing_full=df[df['writing score']==100]['Avg_Score'].count()
math_full=df[df['math score']==100]['Avg_Score'].count()
```

```python
print("Number of students having full reading score is",reading_full)
print("Number of students having full writing score is",writing_full)
print("Number of students having full math score is",math_full)
```

```python
reading_less_20=df[df['reading score']<=20]['Avg_Score'].count()
writing_less_20=df[df['writing score']<=20]['Avg_Score'].count()
math_less_20=df[df['math score']<=20]['Avg_Score'].count()
```

```python
print("Number of students having less than 20 in reading score is",reading_less_20)
print("Number of students having less than 20 in writing score is",writing_less_20)
print("Number of students having less than 20 in math score is",math_less_20)
```
